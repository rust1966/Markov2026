```python
import numpy as np
import pandas as pd

def make_P(m):
    P = np.zeros((m, m))

    for i in range(m - 1):
        P[i, i+1] = 1

    P[m - 1, 0] = 0.5
    P[m - 1, 1] = 0.5

    return P

def find_M(P):
    A = (P > 0).astype(int)
    B = np.eye(P.shape[0], dtype = int)
    M = 0

    while not np.all(B > 0):
        B = (B @ A > 0).astype(int)
        M += 1

    return M

results = []

for m in range(3, 13):
    P = make_P(m)

    M = find_M(P)
    bound = (m - 1) ** 2 + 1

    eigenvals = np.linalg.eigvals(P)
    moduli = np.sort(np.abs(eigenvals))[:: -1]
    second_largest = moduli[1]

    results.append({"m": m, "M": M, "Wielandt bound": bound, "|lambda_2|": second_largest})

results = pd.DataFrame(results)

print(results.to_string(index = False))
```
