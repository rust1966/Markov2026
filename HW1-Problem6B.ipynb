Problem Set-up

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(34)

# Insert the different sample sizes
N_values = np.unique(
    np.logspace(2, 6, 50).astype(int)
)

max_N = N_values[-1]

X = rng.random(max_N)
Y = rng.random(max_N)
Z = rng.random(max_N)

condition_1 = (X ** 2) + (Y ** 2) < Z
condition_2 = X * Y < (Z ** 2)

success = (condition_1 & condition_2)

cumulative_successes = np.cumsum(success)
```

Simulation

```python
estimates = []

for N in N_values:
    number_of_successes = cumulative_successes[N - 1]
    estimate = number_of_successes / N
    estimates.append(estimate)

analytic_answer = 23 * np.pi / 192
```

Show results

```python
print("Monte Carlo estimate:", estimates[-1])
print("Analytic Value:", analytic_answer)

plt.figure(figsize = (8, 5))
plt.semilogx(
    N_values,
    estimates,
    label = "Monte Carlo estimate"
)

plt.axhline(
    y = analytic_answer,
    color = "red",
    linestyle = "--",
    label = r"Analytic Value $23\pi/192$"
    )

plt.xlabel("Sample size N")
plt.ylabel("Estimated Probability")
plt.title("Monte Carlo Estimate of Coupled Inequalities")
plt.grid(True, alpha = 0.3)
plt.legend()
plt.show()
```
